```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_groq import ChatGroq
llm = ChatGroq(model="openai/gpt-oss-120b", streaming=True);
```

```python
question = "What is Gen AI in details?"
```

```python
res = llm.stream(question);
for chunks in res:
    print(chunks.content, end="")
```
