### Here is Tool

```python
from langchain_core.tools import tool
```

```python
@tool
def add_numbers(a: int, b: int) -> int:
    """
    it will return the sum of two numbers a and b.
    Args:
        a (int) : the first number
        b (int) : the second number
    """
    return a + b


@tool
def sub_numbers(a: int, b: int) -> int:
    """
    it will return the difference of two numbers a and b.
    Args:
        a (int) : the first number
        b (int) : the second number
    """
    return a - b
```

```python
ans = add_numbers.invoke({"a": 5, "b": 10});
ans
```

### Here is LLM

```python
from langchain_groq import ChatGroq
llm = ChatGroq(model="openai/gpt-oss-120b");
```

### Now I want to connect Tool + LLM = Agents

```python
from langchain.agents import create_agent
```

```python
agent = create_agent(
    model = llm,
    tools = [add_numbers, sub_numbers],
    system_prompt = "You are a maths teacher and always use tool to answer the questions."
);
```

```python
response =  agent.invoke({"messages": [{"role" : "user", "content" : "What is the sum of 5 and 10 * 5?"}]})
```
