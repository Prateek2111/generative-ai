#### Text Document Loader

```python
from langchain_community.document_loaders import TextLoader
loader = TextLoader("./data/speech.txt")
texts = loader.load()
len(texts) # it show the number of documents loaded - number of pages in the document

onedoc = texts[0]
print(onedoc.page_content)
```

#### Pdf Document Loader

```python
from langchain_community.document_loaders import PyPDFLoader
loader = PyPDFLoader("./data/Application-SLM276M000834.pdf")
texts = loader.load()
len(texts) # it show the number of documents loaded - number of pages in the document

print(texts[0].page_content) # it show the content of the first page of the document
```

#### CSV File Loader

```python
from langchain_community.document_loaders import CSVLoader
loader = CSVLoader(file_path="./data/customers-100.csv", encoding="utf-8")
texts = loader.load()
len(texts) # it show the number of documents loaded - number of pages in the document

print(texts[3].page_content) # in csv file, each row is considered as a document, so it will show the content of the first row of the document
```
