```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.output_parsers import StrOutputParser;
out = StrOutputParser();
```

```python
from langchain_groq import ChatGroq
llm = ChatGroq(model="openai/gpt-oss-120b");
```

```python
from pydantic import BaseModel, Field

class SturcturedOutput(BaseModel):
    name: str = Field(description="Name of the person")
    email : str = Field(description="Email of the person")  
    age: int = Field(description="Age of the person") 
```

```python
structure_llm = llm.with_structured_output(SturcturedOutput);
res = structure_llm.invoke("Please give me my name, email and age from name is Prateek Khare, email is abc@gmail.com and age is 24");
print(res.name); 
```

```python
print(res.model_dump())
```
