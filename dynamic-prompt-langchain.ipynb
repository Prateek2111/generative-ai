```python
from dotenv import load_dotenv
load_dotenv()  # take environment variables from .env.
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI;
from langchain_core.prompts import ChatPromptTemplate;
from langchain_core.output_parsers import StrOutputParser;
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash");
out = StrOutputParser();
```

```python
def transform_upper(result:str):
    return result.upper();
```

```python
prompts = ChatPromptTemplate.from_messages([
    {"role":"system", "content":"You are a translator and translate input in {language} ?"},
    {"role":"user",  "content":"{query}"},
])
```

```python
chains = prompts | llm | out | transform_upper;
```

```python
res = chains.invoke({"language":"Hindi", "query":"What is the capital of France?"});
print(res);
```
