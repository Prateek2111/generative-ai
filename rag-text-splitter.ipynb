```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
texts = "Today, I want to speak from the heart about three pillars that define the future of our nation:" \
"our youth, our education system, and our farmers."
```

#### Document Text Splitter - Load first and then split

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.document_loaders import TextLoader, PyPDFLoader
```

```python
# Text Loader
loader = TextLoader("./data/speech.txt")
texts = loader.load()
```

```python
# PDF Loader
pdf_loader = PyPDFLoader("./data/Application-SLM276M000834.pdf")
pdf = pdf_loader.load()
```

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200);
```

```python
text_chunks = splitter.split_documents(texts);
text_chunks
```

```python
pdf_chunks = splitter.split_documents(pdf);
pdf_chunks
```
